```python
import random

pygame.init()

# 1. Create the game window
CELL = 20
COLS, ROWS = 30, 20
WIDTH, HEIGHT = COLS * CELL, ROWS * CELL

screen = pygame.display.set_mode((WIDTH, HEIGHT))
pygame.display.set_caption("My Snake Game")

clock = pygame.time.Clock()
font = pygame.font.SysFont("Arial", 26)
small_font = pygame.font.SysFont("Arial", 18)

# 2. Define colors
BLACK = (18, 18, 28)
GREEN = (50, 220, 100)
DARK_GREEN = (20, 150, 70)
RED = (255, 80, 80)
WHITE = (245, 245, 245)
GRAY = (45, 45, 58)

# 3. Create food at a random position
def random_food(snake):
    while True:
        food = (random.randrange(COLS), random.randrange(ROWS))
        if food not in snake:
            return food

# 4. Initialize or restart the game
def reset_game():
    snake = [
        (COLS // 2, ROWS // 2),
        (COLS // 2 - 1, ROWS // 2),
        (COLS // 2 - 2, ROWS // 2)
    ]
    direction = (1, 0)
    food = random_food(snake)
    score = 0
    game_over = False
    return snake, direction, food, score, game_over

snake, direction, food, score, game_over = reset_game()
running = True

# 5. Main game loop
while running:

    # Handle keyboard input
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            running = False

        if event.type == pygame.KEYDOWN:

            if event.key == pygame.K_ESCAPE:
                running = False

            if game_over and event.key == pygame.K_SPACE:
                snake, direction, food, score, game_over = reset_game()

            if not game_over:
                if event.key in (pygame.K_UP, pygame.K_w) and direction != (0, 1):
                    direction = (0, -1)

                elif event.key in (pygame.K_DOWN, pygame.K_s) and direction != (0, -1):
                    direction = (0, 1)

                elif event.key in (pygame.K_LEFT, pygame.K_a) and direction != (1, 0):
                    direction = (-1, 0)

                elif event.key in (pygame.K_RIGHT, pygame.K_d) and direction != (-1, 0):
                    direction = (1, 0)

    # Move the snake
    if not game_over:
        head_x, head_y = snake[0]
        new_head = (
            head_x + direction[0],
            head_y + direction[1]
        )

        # Check collision with walls or the snake's body
        body_to_check = snake if new_head == food else snake[:-1]

        if (
            new_head[0] < 0 or new_head[0] >= COLS
            or new_head[1] < 0 or new_head[1] >= ROWS
            or new_head in body_to_check
        ):
            game_over = True

        else:
            snake.insert(0, new_head)

            # Eat food and increase score
            if new_head == food:
                score += 10
                food = random_food(snake)
            else:
                snake.pop()

    # 6. Draw background and grid
    screen.fill(BLACK)

    for x in range(0, WIDTH, CELL):
        pygame.draw.line(screen, GRAY, (x, 0), (x, HEIGHT))

    for y in range(0, HEIGHT, CELL):
        pygame.draw.line(screen, GRAY, (0, y), (WIDTH, y))

    # 7. Draw food
    pygame.draw.rect(
        screen, RED,
        (food[0] * CELL + 2, food[1] * CELL + 2, CELL - 4, CELL - 4),
        border_radius=6
    )

    # 8. Draw the snake
    for i, (x, y) in enumerate(snake):
        color = GREEN if i == 0 else DARK_GREEN

        pygame.draw.rect(
            screen, color,
            (x * CELL + 1, y * CELL + 1, CELL - 2, CELL - 2),
            border_radius=5
        )

    # 9. Display score
    score_text = font.render(f"Score: {score}", True, WHITE)
    screen.blit(score_text, (10, 8))

    # 10. Game-over screen
    if game_over:
        overlay = pygame.Surface((WIDTH, HEIGHT), pygame.SRCALPHA)
        overlay.fill((0, 0, 0, 175))
        screen.blit(overlay, (0, 0))

        title = font.render("GAME OVER", True, RED)
        final_score = small_font.render(
            f"Final Score: {score}", True, WHITE
        )
        restart = small_font.render(
            "Press SPACE to play again", True, WHITE
        )
        quit_text = small_font.render(
            "Press ESC to quit", True, WHITE
        )

        screen.blit(
            title, title.get_rect(center=(WIDTH // 2, HEIGHT // 2 - 45))
        )
        screen.blit(
            final_score, final_score.get_rect(center=(WIDTH // 2, HEIGHT // 2))
        )
        screen.blit(
            restart, restart.get_rect(center=(WIDTH // 2, HEIGHT // 2 + 35))
        )
        screen.blit(
            quit_text, quit_text.get_rect(center=(WIDTH // 2, HEIGHT // 2 + 60))
        )

    pygame.display.flip()
    clock.tick(10)

pygame.quit()
```
